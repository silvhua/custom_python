# Take Gifts From the Richest Pile

You are given the root of a binary tree, return the preorder traversal of its nodes' values.

* [LeetCode link]()
* [NeetCode link](https://neetcode.io/problems/take-gifts-from-the-richest-pile/question?list=neetcode150)

You are given an integer array gifts denoting the number of gifts in various piles. Every second, you do the following:

* Choose the pile with the maximum number of gifts.
* If there is more than one pile with the maximum number of gifts, choose any.
* Reduce the number of gifts in the pile to the floor of the square root of the original number of gifts in the pile.
Return the number of gifts remaining after k seconds.

Example 1:
```
Input: gifts = [25,64,9,4,100], k = 4

Output: 29
```
Explanation :
The gifts are taken in the following way:

* In the first second, the last pile is chosen and 10 gifts are left behind.
* Then the second pile is chosen and 8 gifts are left behind.
* After that the first pile is chosen and 5 gifts are left behind.
* Finally, the last pile is chosen again and 3 gifts are left behind.
* The final remaining gifts are [5,8,9,4,3], so the total number of gifts remaining is 29.

Example 2:
```
Input: gifts = [1,1,1,1], k = 4

Output: 4
```
Constraints:
* 1 <= gifts.length <= 1000
* 1 <= gifts[i] <= 1,000,000,000
* 1 <= k <= 1000

## Skeleton code
```python
class Solution:
    def pickGifts(self, gifts: List[int], k: int) -> int:
```

## Iteration 1
* Time: $ O() $
* Space: $ O() $

````python
from math import sqrt, floor

class Solution:
    def pickGifts(self, gifts, k) -> int:
        piles = gifts
        while k > 0:
            piles = Heap.heapify_max(piles)
            pile = piles[1]
            pile = floor(sqrt(pile))
            piles[1] = pile
            k -= 1
        return sum(piles)

class Heap:
    def __init__(self):
        pass

    @staticmethod
    def heapify_max(array):
        array = [0] + array
        i = (len(array) - 1) // 2
        heap = Heap.percolate_max(array, i)
        return heap

    @staticmethod 
    def percolate_max(array, i):
        messages = []
        messages.append(f'array: {array}')
        while i > 0:
            current = array[i]
            right_child_index = i * 2 + 1
            left_child_index = i * 2
            messages.append(f'\ti: {i}\tcurrent: {current}\tleft child index: {left_child_index}\tright child index: {right_child_index}')
            if left_child_index < len(array):
                messages.append(f"\t\tleft child: {array[left_child_index]}")
            if right_child_index < len(array):
                messages.append(f'\t\tright child: {array[right_child_index]}')
            if right_child_index < len(array) and array[right_child_index] > array[left_child_index] and current < array[right_child_index]:
                messages.append("\t\tSwap with right child")
                array[i] = array[right_child_index]
                array[right_child_index] = current
                i = i * 2 + 1
            elif left_child_index < len(array) and current < array[left_child_index]:
                messages.append("\t\tSwap with left child")
                array[i] = array[left_child_index]
                array[left_child_index] = current
                i = i * 2
            else:
                messages.append("\t\tNo swap")
                i -= 1
        return array

cases = {
    1: {
        "gifts": [25,64,9,4,100], 
        "k": 4,
        "Output": 29
    },
    2: {
        "gifts": [1,1,1,1], 
        "k": 4,
        "Output": 4
    }
}

correct = []
solution = Solution()

for case_number in cases:
    case = cases[case_number]
    print(f"*** Case {case_number} ***")
    for key in case.keys():
        print(f"{key}: {case[key]}", end="\t")
    print("")
    
    gifts = case["gifts"]
    k = case["k"]
    Output = case["Output"]
    result = solution.pickGifts(gifts, k)
    correct_for_case = result == Output
    print(f"Result: {result}")
    correct.append(correct_for_case)
    print(f"Answer is correct: {correct_for_case}\n\n")
print(f"\n***** All cases correct: {sum(correct) == len(correct)} *****")
````
