# Combinations

* [LeetCode link]()
* [NeetCode link](https://neetcode.io/problems/combinations/question?list=neetcode150)

You are given two integers `n` and `k`, return all possible combinations of `k` numbers chosen from the range `[1, n]`.

You may return the answer in any order.

Example 1:
```
Input: n = 3, k = 2

Output: [
    [1,2],
    [1,3],
    [2,3]
]
```
Example 2:
```
Input: n = 3, k = 3

Output: [[1,2,3]]
```
Constraints:
`1 <= k <= n <= 20`

## Skeleton code
```python
class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
```

## Iteration 1
* Time: $ O() $
* Space: $ O() $

````python
from typing import List

class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
        choose = Combination(n, k)
        result = choose.get_all_sets()
        return result

class Combination:
    def __init__(self, n, k):
        self.n = n
        self.k = k
        self.sets = []

    def get_all_sets(self):
        for i in range(1, self.n + 1): 
            print(f'first value: {i}')
            self.choose(i, [])
        return self.sets

    def choose(self, current_value, current_set):
        print(f'\tvalue: {current_value}')
        current_set.append(current_value)
        if len(current_set) == self.k:
            self.sets.append(current_set.copy())
            print(f'\t\tCombination complete: {current_set}')
            return
        for i in range(current_value + 1, self.n + 1):
            self.choose(i, current_set)


cases = {
    1: {
            "n": 3, 
            "k": 2,
            "Output": [
                [1,2],
                [1,3],
                [2,3]
            ]
        },
        2: {
            "n": 3, 
            "k": 3,
            "Output": [[1,2,3]]
        }
}

correct = []
solution = Solution()

for case_number in cases:
    case = cases[case_number]
    print(f"*** Case {case_number} ***")
    for key in case.keys():
        print(f"{key}: {case[key]}", end="\t")
    print("")
    
    Output = case["Output"]
    n = case["n"]
    k = case["k"]
    result = solution.combine(n, k)
    correct_for_case = result == Output
    print(f"Result: {result}")
    correct.append(correct_for_case)
    print(f"Answer is correct: {correct_for_case}\n\n")
print(f"\n***** All cases correct: {sum(correct) == len(correct)} *****")
````
