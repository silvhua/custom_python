# Create all possible subsets

## Iteration 1
* Time: $ O() $
* Space: $ O() $

```python
class Subsets:
    def __init__(self, elements):
        self.elements = elements
        self.subsets = []
        self.messages = []

    def get_all_subsets(self):
        current = []
        for i in range(len(self.elements)):
            self.helper(i, current)
        n_options = len(self.subsets)
        self.messages.append(f'\tNumber of options: {n_options} (expected: {n_options == len(elements) ^ 2})')
        self.messages.append(f'\tOptions: ')
        for index, subset in enumerate(self.subsets):
            self.messages.append(f'\t\t{index}: '+', '.join([f'{val}' for val in subset]))
        print('\n'.join([mesage for mesage in self.messages]))

    def helper(self, i, current):
        # self.messages.append(f'i: {i}')
        current = current.copy()
        if i == len(self.elements):
            self.subsets.append(current)
            self.messages.append(f'\tLast element reached: ' + f'{[val for val in current]}')
            return

        # Option 1: Exclude this number
        self.helper(i + 1, current)

        # Option 2: include this number
        current.append(elements[i])
        self.helper(i + 1, current)

elements = ['a', 'b', 'c']
instance = Subsets(elements)
instance.get_all_subsets()
```

# Use what is in NeetCode

```python
def subsetsWithoutDuplicates(nums):
    subsets, curSet = [], []
    helper(0, nums, curSet, subsets)
    print(f'Number of subsets: {len(subsets)}')
    return subsets

def helper(i, nums, curSet, subsets):
    print(f'i: {i}')
    if i >= len(nums):
        print(f'\tsubset: {curSet}')
        subsets.append(curSet.copy())
        return

    # decision to include nums[i]
    curSet.append(nums[i])
    helper(i + 1, nums, curSet, subsets)
    curSet.pop()

    # decision NOT to include nums[i]
    helper(i + 1, nums, curSet, subsets)

nums = ['a', 'b', 'c']
subsetsWithoutDuplicates(nums)
```

## With duplicate elements

```python
class Subsets:
    def __init__(self, elements):
        self.elements = elements
        self.subsets = []
        self.messages = []

    def get_all_subsets_with_duplicates(self):
        current = []
        self.elements_sorted = sorted(self.elements.copy())
        self.helper2(0, current)
        n_options = len(self.subsets)
        self.messages.append(f'\tNumber of options: {n_options}')
        self.messages.append(f'\tOptions: ')
        for index, subset in enumerate(self.subsets):
            self.messages.append(f'\t\t{index}: '+', '.join([f'{val}' for val in subset]))
        print('\n'.join([mesage for mesage in self.messages]))

    def helper2(self, i, current):
        if i == len(self.elements_sorted):
            self.subsets.append(current.copy())
            self.messages.append(f'\tLast element reached: ' + f'{[val for val in current]}')
            return

        # Option 1: include this number
        # self.messages.append(f'\tOption 1')
        current.append(self.elements_sorted[i])
        self.helper2(i + 1, current)

        # Option 2: Exclude this number
        # self.messages.append(f'\tOption 2')
        current.pop()
        while i + i < len(self.elements_sorted) and self.elements[i + 1] == self.elements[i]:
            i += 1
        self.helper2(i + 1, current)

    def get_all_subsets(self):
        current = []
        self.helper(0, current)
        n_options = len(self.subsets)
        self.messages.append(f'\tNumber of options: {n_options} (expected: {n_options == 2**len(elements)})')
        self.messages.append(f'\tOptions: ')
        for index, subset in enumerate(self.subsets):
            self.messages.append(f'\t\t{index}: '+', '.join([f'{val}' for val in subset]))
        print('\n'.join([mesage for mesage in self.messages]))

    def helper(self, i, current):
        if i == len(self.elements):
            self.subsets.append(current.copy())
            # self.messages.append(f'\tLast element reached: ' + f'{[val for val in current]}')
            return

        # Option 1: include this number
        # self.messages.append(f'\tOption 1')
        current.append(elements[i])
        self.helper(i + 1, current)

        # Option 2: Exclude this number
        # self.messages.append(f'\tOption 2')
        current.pop()
        self.helper(i + 1, current)


elements = ['a', 'b', 'b', 'c']
instance = Subsets(elements)
instance.get_all_subsets_with_duplicates()
```
