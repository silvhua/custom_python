# Recursive

```python
# Time: O(n^2 * n!)
def permutationsRecursive(nums):
    return helper(0, nums)

def helper(i, nums):
    print(f'i: {i}')
    if i == len(nums):
        return [[]]

    resPerms = []
    perms = helper(i + 1, nums)
    print(f'\tPermutations for i = {i}: {[perm for perm in perms]}')
    for p in perms:
        print(f'\t\tPermutation: {p}')
        for j in range(len(p) + 1):
            pCopy = p.copy()
            pCopy.insert(j, nums[i])
            resPerms.append(pCopy)
    return resPerms

nums = [1, 2, 3]
result = permutationsRecursive(nums)
print(f'Result: {[p for p in result]} (n = {len(result)})')
```
