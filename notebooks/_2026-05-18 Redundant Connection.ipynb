# Redundant Connection

* [LeetCode link]()
* [NeetCode link](https://neetcode.io/problems/redundant-connection/question?list=neetcode150)

You are given a connected undirected graph with `n` nodes labeled from `1` to `n`. Initially, it contained no cycles and consisted of `n-1` edges.

We have now added one additional edge to the graph. The edge has two different vertices chosen from `1` to `n`, and was not an edge that previously existed in the graph.

The graph is represented as an array `edges` of length `n` where `edges[i] = [ai, bi]` represents an edge between nodes ai and bi in the graph.

Return an edge that can be removed so that the graph is still a connected non-cyclical graph. If there are multiple answers, return the edge that appears last in the input `edges`.

Example 1:
```
Input: edges = [[1,2],[1,3],[3,4],[2,4]]

Output: [2,4]
```
Example 2:
```
Input: edges = [[1,2],[1,3],[1,4],[3,4],[4,5]]

Output: [3,4]
```

Constraints:
* n == edges.length
* 3 <= n <= 100
* 1 <= edges[i][0] < edges[i][1] <= edges.length
There are no repeated edges and no self-loops in the input.

````python
class Solution:
    def findRedundantConnection(self, edges):
        """
        Find the first edge where each node has the same non-self
        """
        for index in range(len(edges) - 1, -1, -1):
            edge = edges[index]
            self.messages.append(f"index: {index}\tnode: {edge}")
            same_parent = self.compare_parents(edge[0], edge[1])
            if same_parent == True:
                return edge


    def __init__(self, nodes):
        self.parents = {}
        self.direct_parent = {}
        self.height = {}
        for node in nodes:
            self.parents[node] = node
            self.height[node] = 0
        self.nodes = nodes 
        self.messages = []
        self.info_messages = []

    def find(self, value):
        if value != self.parents[value]:
            self.parents[value] = self.find(self.parents[value])
        self.messages.append(f'Parent of {value} is {self.parents[value]}')
        return self.parents[value]

    # def find(self, value):
    #     """
    #     Find the direct parent (no path compression)
    #     """
    #     parent = self.parents[value]
    #     return parent

    def compare_parents(self, node1, node2):
        parent1 = self.find(node1)
        parent2 = self.find(node2)
        return parent1 == parent2 and parent2 != node2 and parent1 != node1

    def union(self, node1, node2):
        parent1 = self.find(node1)
        parent2 = self.find(node2)
        if parent1 == parent2:
            return False
        height1 = self.height[parent1]
        height2 = self.height[parent2] 
        if height1 > height2:
            self.parents[parent2] = parent1
            self.messages.append(f'Setting node2 parent to parent1')
        else:
            self.parents[parent1] = parent2
            self.messages.append(f'Setting node1 parent to parent2')
            if height2 == height1:
                self.messages.append(f'\tIncrementing parent2 height by 1')
                height2 += 1

        return True


cases = {
    1: {
        "edges": [[1,2],[1,3],[3,4],[2,4]],
        "Output": [2,4]
    },
    2: {
        "edges": [[1,2],[1,3],[1,4],[3,4],[4,5]],
        "Output": [3,4]
    }
}

correct = []

for case_number in cases:
    case = cases[case_number]
    print(f"*** Case {case_number} ***")
    for key in case.keys():
        print(f"{key}: {case[key]}", end="\t")
    print("")
    
    edges = case["edges"]
    node_values = []
    for pair in edges:
        node_values += pair
    node_values = list(set(node_values))
    print(f'Node values: {node_values}')
    Output = case["Output"]
    solution = Solution(node_values)
    for pair in edges:
        solution.union(pair[0], pair[1])
    solution.findRedundantConnection(edges)
    print('\t' + '\n\t'.join(solution.info_messages))
    # print('\n\t'.join(solution.messages))
    for node in solution.parents:
        print(f'\tnode {node} parent: {solution.parents[node]}')
    result = solution.findRedundantConnection(edges)
    correct_for_case = result == Output
    print(f"Result: {result}")
    correct.append(correct_for_case)
    print(f"Answer is correct: {correct_for_case}\n\n")
print(f"\n***** All cases correct: {sum(correct) == len(correct)} *****")
````

````python
class Solution:

    def findRedundantConnection(self, edges):
        """
        Find the first edge where each node has the same non-self direct parent
        """
        node_values = []
        edges_forming_cyle = []
        for pair in edges:
            node_values += pair
        node_values = list(set(node_values))
        self.union_find = UnionFind(node_values)
        for pair in edges:
            can_union = self.union_find.union(pair[0], pair[1])
            if can_union == False:
                edges_forming_cyle.append(pair)
        return edges_forming_cyle[-1]

class UnionFind:

    def __init__(self, nodes):
        print(f'Node values: {nodes}')
        self.parents = {}
        self.direct_parent = {}
        self.height = {}
        for node in nodes:
            self.parents[node] = node
            self.height[node] = 0
            self.direct_parent[node] = node
        self.nodes = nodes 
        self.messages = []
        self.info_messages = []

    def find(self, value, first_iteration = True):
        if value != self.parents[value]:
            self.parents[value] = self.find(self.parents[value], False)
            if first_iteration == True:
                self.direct_parent[value] = self.parents[value]
        self.messages.append(f'Parent of {value} is {self.parents[value]}\tDirect parent is {self.direct_parent[value]}')
        return self.parents[value]

    def compare_parents(self, node1, node2):
        direct_parent1 = self.direct_parent[node1]
        direct_parent2 = self.direct_parent[node2]
        return direct_parent1 == direct_parent2 and direct_parent2 != node2 and direct_parent1 != node1

    def union(self, node1, node2):
        self.messages.append(f'*** Union of {node1} and {node2}')
        parent1 = self.find(node1)
        parent2 = self.find(node2)
        if parent1 == parent2:
            self.messages.append(f'\tCycle detected; cannot perform union')
            return False
        height1 = self.height[parent1]
        height2 = self.height[parent2] 
        if height1 > height2 or height2 == height1:
            self.parents[parent2] = parent1
            self.direct_parent[node2] = node1
            if height2 == height1:
                self.messages.append(f'\tHeight 1 = Height 2')
                self.messages.append(f'\t\tIncrementing parent1 (`{parent1}`) height by 1')
                self.height[parent1] += 1
            else:
                self.messages.append(f'\tHeight 1 > Height 2')
            self.messages.append(f'\tSetting node2 parent, `{parent2}`, to parent1, `{parent1}`, node `{node2}` direct parent to `{node1}`')
        else:
            self.messages.append(f'\tHeight 2 > Height 1')
            self.parents[parent1] = parent2
            self.direct_parent[node1] = node2
            self.messages.append(f'\tSetting node1 parent, `{parent1}`, to parent2, `{parent2}`, node `{node1}` direct parent to `{node2}`')

        return True


cases = {
    1: {
        "edges": [[1,2],[1,3],[3,4],[2,4]],
        "Output": [2,4]
    },
    2: {
        "edges": [[1,2],[1,3],[1,4],[3,4],[4,5]],
        "Output": [3,4]
    },
    3: {
        "edges": [[1,2],[2,3],[3,4],[1,4],[1,5]],
        "Output": [1,4]
    }
}

correct = []

for case_number in cases:
    case = cases[case_number]
    print(f"*** Case {case_number} ***")
    for key in case.keys():
        print(f"{key}: {case[key]}", end="\t")
    print("")
    
    edges = case["edges"]
    Output = case["Output"]
    solution = Solution()
    result = solution.findRedundantConnection(edges)
    # result = None
    # print('\t' + '\n\t'.join(solution.info_messages))
    for node in solution.union_find.parents:
        print(f'\tnode {node} parent: {solution.union_find.parents[node]}')
    print('\t' + '\n\t'.join(solution.union_find.messages))
    correct_for_case = result == Output
    print(f"Result: {result}")
    correct.append(correct_for_case)
    print(f"Answer is correct: {correct_for_case}\n\n")
print(f"\n***** All cases correct: {sum(correct) == len(correct)} *****")
````
