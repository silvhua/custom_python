# My Calendar I

* [LeetCode link]()
* [NeetCode link](https://neetcode.io/problems/my-calendar-i/question?list=neetcode150)

You are implementing a program to use as your calendar. We can add a new event if adding the event will not cause a double booking.

A double booking happens when two events have some non-empty intersection (i.e., some moment is common to both events.).

The event can be represented as a pair of integers `startTime` and `endTime` that represents a booking on the half-open interval `[startTime, endTime)`, the range of real numbers `x` such that `startTime <= x < endTime`.

Implement the `MyCalendar` class:

* `MyCalendar()` Initializes the calendar object.
* `boolean book(int startTime, int endTime)` Returns `true` if the event can be added to the calendar successfully without causing a double booking. Otherwise, return `false` and do not add the event to the calendar.
Example 1:
```
Input:
["MyCalendar", "book", "book", "book"]
[[], [10, 20], [15, 25], [20, 30]]

Output:
[null, true, false, true]
Explanation:
MyCalendar myCalendar = new MyCalendar();
myCalendar.book(10, 20); // return True
myCalendar.book(15, 25); // return False, It can not be booked because time 15 is already booked by another event.
myCalendar.book(20, 30); // return True, The event can be booked, as the first event takes every time less than 20, but not including 20.
```
Constraints:

* `0 <= start < end <= 1,000,000,000`
* At most 1000 calls will be made to `book`.

* Time: $ O() $

````python
class SegmentTree:
    def __init__(self, elements):
        """
        Each node represents an appointment and has the start time and end time 
        """
        self.min = min(elements) if len(elements) > 0 else 0
        self.max = max(elements) if len(elements) > 0 else 0
        self.elements = set(elements)
        self.left_tree = None
        self.right_tree = None
        print(f'\tCreating new SegmentTree with min {self.min}, max {self.max}, elements {self.elements}')

    @staticmethod
    def build(L, R):
        if L == M:
            return SegmentTree(elements[L])
        M = (R - L )// 2
        elements = [i for i in range(L, R)]
        root = SegmentTree(elements)
        root.right = SegmentTree(elements[L, M])
        root.left = SegmentTree(elements[M+1, R])
        return root

    def add(self, root, startTime, endTime):
        elements = [i for i in range(startTime, endTime)]
        print(f'Attemping to add appointment with start time {startTime} and end time {endTime}. Elements: {elements}')
        if root.elements.intersection(elements):
            print(f'\tIntersection found')
            return None
        print(f'\tRoot min: {root.min}; root max: {root.max}')
        if root.min == root.max: # no appointments yet
            root = SegmentTree(elements)
            print(f'\tCreating root.')
        M = len(elements) // 2 # middle index
        # SegmentTree.add(root, startTime, elements[M])
        # SegmentTree.add(root, elements[M+1], endTime)
        root.left_tree = SegmentTree(elements[:M])
        root.right_tree = SegmentTree(elements[M+1:])
        return root



class MyCalendar:
    
    def __init__(self):
        self.tree = SegmentTree([])
        return None

    def book(self, startTime: int, endTime: int) -> bool:
        root = self.tree.add(self.tree, startTime, endTime)
        if root is not None:
            self.tree = root
            return True
        else:
            return False

# Your MyCalendar object will be instantiated and called as such:
# obj = MyCalendar()
# param_1 = obj.book(startTime,endTime)


cases = {
    1: {
        "Input": [
            [], 
            [10, 20], 
            [15, 25], 
            [20, 30]
            ],
        "Output": [None, True, False, True]
    }
}

correct = []

for case_number in cases:
    case = cases[case_number]
    print(f"*** Case {case_number} ***")
    for key in case.keys():
        print(f"{key}: {case[key]}", end="\t")
    print("")
    
    Output = case["Output"]
    input = case["Input"]
    obj = MyCalendar()
    result = [None]
    for appointment in input[1:]:
        result.append(obj.book(appointment[0], appointment[1]))
    correct_for_case = result == Output
    print(f"Result: {result}")
    correct.append(correct_for_case)
    print(f"Answer is correct: {correct_for_case}\n\n")
print(f"\n***** All cases correct: {sum(correct) == len(correct)} *****")
````
