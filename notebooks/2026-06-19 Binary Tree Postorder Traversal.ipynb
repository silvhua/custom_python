# Binary Tree Preorder Traversal

You are given the root of a binary tree, return the preorder traversal of its nodes' values.

* [LeetCode link]()
* [NeetCode link](https://neetcode.io/problems/binary-tree-preorder-traversal/question)

Example 1:
```
Input: root = [1,2,3,4,5,6,7]

Output: [1,2,4,5,3,6,7]
```
Example 2:
```
Input: root = [1,2,3,null,4,5,null]

Output: [1,2,4,3,5]
```
Example 3:
```
Input: root = []

Output: []
```

Constraints:
* 0 <= number of nodes in the tree <= 100
* -100 <= Node.val <= 175

Follow up: Recursive solution is trivial, could you do it iteratively?

## Skeleton code
```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def postorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
```

## Iteration 1: 
* Time: $ O(n) $
* Space: $ O(2n) $

````python
from typing import Optional, List
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def postorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        result = []
        stack = [root]
        visited = [False]
        while stack:
            pointer = stack.pop()
            if not pointer:
                continue
            node_visited = visited.pop()
            if node_visited == True and pointer.val is not None:
                result.append(pointer.val)
                
            elif node_visited == False:
                stack.append(pointer)
                visited.append(True)
                if pointer.right:
                    stack.append(pointer.right)
                    visited.append(False)
                if pointer.left:
                    stack.append(pointer.left)
                    visited.append(False)

        return result

    def buildTree(self, nodes):
        # Treat the tree as a heap
        nodes = [None] + nodes
        self.tree = self.createNode(1, nodes)
        return self.tree

    def createNode(self, index, nodes):
        if index >= len(nodes):
            # print(f"\tindex exceeded\n\t\tindex: {index}\t# of nodes: {len(nodes)}")
            return None
        left_child_index = index * 2
        
        right_child_index = left_child_index + 1
        val = nodes[index]
        # print(f"\tcreateNode index: {index}\tvalue: {val}")
        # print(f"\t\tleft child index: {left_child_index}\tright child index: {right_child_index}")
        if val:
            node = TreeNode(
                val,
                self.createNode(left_child_index, nodes),
                self.createNode(right_child_index, nodes)
                )
        else:
            node = TreeNode(val, None, None)
        return node
    
    def printNodes(self, root, nodes_list = []):
        if not root:
            return nodes_list
        nodes_list.append(root.val)
        self.addToNodesList(root, nodes_list)
        print(f"Tree nodes: {[val for val in nodes_list]}")
        return nodes_list

    def addToNodesList(self, node, nodes_list, index=1):
        if not node:

            return
        if node.left:
            nodes_list.append(node.left.val)
        if node.right:
            nodes_list.append(node.right.val)

        self.addToNodesList(node.left, nodes_list)
        self.addToNodesList(node.right, nodes_list)
cases = {
    1: {
       "root": [1,2,3,4,5,6,7],
        "Output": [4,5,2,6,7,3,1]
    },
    2: {
        "root": [1,2,3,None,4,5,None],
        "Output": [4,2,5,3,1]
    },
    3: {
        "root": [],
        "Output": []
    }
}

correct = []
solution = Solution()

for case_number in cases:
    case = cases[case_number]
    print(f"*** Case {case_number} ***")
    for key in case.keys():
        print(f"{key}: {case[key]}", end="\t")
    print("")
    
    root = case["root"]
    Output = case["Output"]
    tree = solution.buildTree(root)

    result = solution.postorderTraversal(tree)
    correct_for_case = result == Output
    print(f"Result: {result}")
    correct.append(correct_for_case)
    print(f"Answer is correct: {correct_for_case}\n\n")
print(f"\n***** All cases correct: {sum(correct) == len(correct)} *****")
````
