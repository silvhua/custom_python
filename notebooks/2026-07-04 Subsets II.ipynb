# Subsets II

* [LeetCode link]()
* [NeetCode link](https://neetcode.io/problems/subsets-ii/question?list=neetcode150)

You are given an array `nums` of integers, which may contain duplicates. Return all possible subsets.

The solution must not contain duplicate subsets. You may return the solution in any order.

Example 1:
```
Input: nums = [1,2,1]

Output: [[],[1],[1,2],[1,1],[1,2,1],[2]]
```
Example 2:
```
Input: nums = [7, 7]

Output: [[],[7], [7,7]]
```
Constraints:

* 1 <= nums.length <= 10
* -10 <= nums[i] <= 10

## Skeleton code
```python
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
```

## Iteration 1
* Time: $ O(n 2^n) $
* Space: $ O(n) $

- Runtime 55ms | Beats 8.35% 
- Memory 8.9 MB | Beats 13.91%

````python
class Solution:
    def subsetsWithDup(self, nums):
        instance = Subsets(nums)
        instance.get_all_subsets_with_duplicates()
        return instance.subsets

class Subsets:
    def __init__(self, elements):
        self.elements = elements
        self.subsets = []
        self.messages = []

    def get_all_subsets_with_duplicates(self):
        current = []
        self.elements_sorted = sorted(self.elements.copy())
        self.helper2(0, current)
        n_options = len(self.subsets)
        self.messages.append(f'\tNumber of options: {n_options}')
        self.messages.append(f'\tOptions: ')
        for index, subset in enumerate(self.subsets):
            self.messages.append(f'\t\t{index}: '+', '.join([f'{val}' for val in subset]))
        print('\n'.join([mesage for mesage in self.messages]))

    def helper2(self, i, current):
        if i == len(self.elements_sorted):
            self.subsets.append(current.copy())
            self.messages.append(f'\tLast element reached: ' + f'{[val for val in current]}')
            return

        # Option 1: include this number
        # self.messages.append(f'\tOption 1')
        current.append(self.elements_sorted[i])
        self.helper2(i + 1, current)

        # Option 2: Exclude this number
        # self.messages.append(f'\tOption 2')
        current.pop()
        while i + i < len(self.elements_sorted) and self.elements_sorted[i + 1] == self.elements_sorted[i]:
            i += 1
        self.helper2(i + 1, current)

    def get_all_subsets(self):
        current = []
        self.helper(0, current)
        n_options = len(self.subsets)
        self.messages.append(f'\tNumber of options: {n_options} (expected: {n_options == 2**len(self.elements)})')
        self.messages.append(f'\tOptions: ')
        for index, subset in enumerate(self.subsets):
            self.messages.append(f'\t\t{index}: '+', '.join([f'{val}' for val in subset]))
        print('\n'.join([mesage for mesage in self.messages]))

    def helper(self, i, current):
        if i == len(self.elements):
            self.subsets.append(current.copy())
            # self.messages.append(f'\tLast element reached: ' + f'{[val for val in current]}')
            return

        # Option 1: include this number
        # self.messages.append(f'\tOption 1')
        current.append(self.elements[i])
        self.helper(i + 1, current)

        # Option 2: Exclude this number
        # self.messages.append(f'\tOption 2')
        current.pop()
        self.helper(i + 1, current)

cases = {
    1: {
        "nums": [1,2,1],
        "Output": [[],[1],[1,2],[1,1],[1,2,1],[2]] 
    },
    2: {
        "nums": [7, 7],
        "Output": [[],[7], [7,7]]
    },
    3: {
        "nums": [0],
        "Output": [[0],[]]
    }
}

correct = []
solution = Solution()

for case_number in cases:
    case = cases[case_number]
    print(f"*** Case {case_number} ***")
    for key in case.keys():
        print(f"{key}: {case[key]}", end="\t")
    print("")
    
    Output = case["Output"]
    nums = case["nums"]
    result = solution.subsetsWithDup(nums)
    correct_for_case = result == Output
    print(f"Result: {result}")
    correct.append(correct_for_case)
    print(f"Answer is correct: {correct_for_case}\n\n")
print(f"\n***** All cases correct: {sum(correct) == len(correct)} *****")
````
